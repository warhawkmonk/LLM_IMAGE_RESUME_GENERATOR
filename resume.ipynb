# Call the Resume Generation API

````python
import requests
def consume_llm_api(prompt,image):
    """
    Sends a prompt to the LLM API and processes the streamed response.
    """
    url = "https://www.aatu-zhaatu.xyz/api/llm-response"
    headers = {"Content-Type": "application/json"}
    payload = {"prompt": prompt,"image_b64":image,"image_understanding":True}


    response = requests.post(url, json=payload, headers=headers)

    return response.text
        
````

# Encode the Reference Image

````python
import base64

def image_to_base64(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

b64_str = image_to_base64("image_folder\\res-2.jpg")
````

````python
value = consume_llm_api("""
You are a professional resume-writing assistant. Convert attached image design into a single HTML resume.

STRICT OUTPUT RULES:

 - Output ONLY HTML. No explanations, no comments, no code fences, no text before or after the resume.

 - Copy the image structure exactly, in the given order.

 - Match the visual hierarchy using HTML only as same as possible.

 - Ensure extreme precision in design and formatting (important).

 - Give importance to design color font size alignment. 

 - below code is going to use to convert html into pdf so must create so that it looks great.

        from weasyprint import HTML, CSS
        PAGE_WIDTH_PX = 1000
        PAGE_HEIGHT_PX = 1400 
        MARGIN_MM = 0
        SCALE = 1.0
        
        INPUT_FILE = 'index.html'
        OUTPUT_FILE = 'output.pdf'
        
        override_css = CSS(string=f'''
            @page {{
                size: {PAGE_WIDTH_PX}px {PAGE_HEIGHT_PX}px;
                margin: 0 !important;
            }}

            html {{
                margin: 0 !important;
                padding: 0 !important;
            }}

            body {{
                margin: 0 !important;
                padding: 0 !important;
                background: white !important;
                transform: scale({SCALE});
                transform-origin: top left;
            }}

            body > div,
            .container,
            .wrapper,
            .resume,
            .resume-container,
            .resume-wrapper,
            .page,
            .page-container,
            .a4,
            .sheet {{
                margin: 0 !important;
                padding: 0 !important;
                max-width: 100% !important;
                width: 100% !important;
            }}

            .sidebar,
            .left,
            .left-column,
            .left-panel {{
                margin: 0 !important;
            }}
        ''')

        HTML(INPUT_FILE).write_pdf(OUTPUT_FILE, stylesheets=[override_css])
""",b64_str)
````
