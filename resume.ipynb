````python
import requests
def consume_llm_api(prompt,image):
    """
    Sends a prompt to the LLM API and processes the streamed response.
    """
    url = "http://127.0.0.1:6000/api/llm-response"
    headers = {"Content-Type": "application/json"}
    payload = {"prompt": prompt,"image_b64":image,"image_understanding":True}


    response = requests.post(url, json=payload, headers=headers)
    result = response.json()
    return result.get("text", "")
        
````

````python
import base64

def image_to_base64(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

b64_str = image_to_base64("res-2.jpg")
````

````python
from weasyprint import HTML, CSS

# ==== CUSTOMIZE THESE ====
PAGE_WIDTH_PX = 1000
PAGE_HEIGHT_PX = 1400 
MARGIN_MM = 0
SCALE = 1.0
INPUT_FILE = 'index.html'
OUTPUT_FILE = 'output.pdf'
# ==========================

override_css = CSS(string=f'''
    @page {{
        size: {PAGE_WIDTH_PX}px {PAGE_HEIGHT_PX}px;
        margin: 0 !important;
    }}

    html {{
        margin: 0 !important;
        padding: 0 !important;
    }}

    body {{
        margin: 0 !important;
        padding: 0 !important;
        background: white !important;
        transform: scale({SCALE});
        transform-origin: top left;
    }}

    /* Kill margin/padding on common wrapper/container class names */
    body > div,
    .container,
    .wrapper,
    .resume,
    .resume-container,
    .resume-wrapper,
    .page,
    .page-container,
    .a4,
    .sheet {{
        margin: 0 !important;
        padding: 0 !important;
        max-width: 100% !important;
        width: 100% !important;
    }}

    /* Ensure sidebar/left column has no outer gap, only its own internal padding */
    .sidebar,
    .left,
    .left-column,
    .left-panel {{
        margin: 0 !important;
    }}
''')

HTML(INPUT_FILE).write_pdf(OUTPUT_FILE, stylesheets=[override_css])
````
